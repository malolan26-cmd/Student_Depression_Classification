```python
# Data from https://www.kaggle.com/datasets/mkhoirulmustaqim/student-depression-detection-dataset?resource=download
import pandas as pd
```

```python
df = pd.read_csv('Student_Depression_Dataset.csv')
```

```python
# Encode gender, family background, relationship status, marital status, living style, friend circle and health_condition and depression (target)
# Drop column on home division
categorical_features = [
    'Gender',
    'Family_background',
    'Relationship_status',
    'Marital_status',
    'Living_style',
    'friend_circle',
    'Health_condition'
]

df = df.drop(columns=['Home_division'])
```

```python
# Encode the binary target as 0/1.
df['Depression'] = df['Depression'].map({'No': 0, 'Yes': 1}).astype('int64')

# One-hot encode categorical predictors and avoid redundant dummy columns.
df = pd.get_dummies(
    df,
    columns=categorical_features,
    drop_first=True,
    dtype='int64'
)
```

```python
X = df.drop(columns = 'Depression')
y = df['Depression']
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import root_mean_squared_error
from sklearn.neighbors import KNeighborsClassifier
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=10, shuffle=True, stratify=y)
```

```python
numeric = ["Age", "Current_semester", "CGPA", "Cost_without_study"]
```

```python
preprocessing = ColumnTransformer(
    [('numeric', StandardScaler(), numeric)], remainder = 'passthrough')
```

```python
model = make_pipeline(
    preprocessing,
    KNeighborsClassifier(n_neighbors=5)
)
```

```python
kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=10)
```

```python
scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv = kfold,
    scoring = 'roc_auc'
)
```

```python
print(scores.mean(), scores.std())
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test, y_pred)
```
